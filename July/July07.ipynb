# Uses 'CaptureFlagBoard1.csv'

```python
import pandas as pd
import numpy as np 
import plotly.express as px

nums = []
for i in range(16, 104):
    nums.append(i)

data = pd.read_csv("CaptureFlagBoard1.csv")
data = data.drop(data.columns[nums], axis=1)
data
```

```python
current_x = list(data['CurrentX'])
current_y = list(data['CurrentY'])
destination_x = list(data['DestinationX'])
destination_y = list(data['DestinationY'])
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

x = np.append(current_x, destination_x)
y = np.append(current_y, destination_y)

fig, ax = plt.subplots()
ax.scatter(x, y, c = "lime", edgecolor = 'black', s = 50)

#ax.set_title('Scatter plot')
#ax.set_xlabel('x value')
#ax.set_ylabel('y value')

plt.show()
```
