# Uses 'JointEvaluation.csv' and 'SingleEvaluation.csv'

```python
import csv
from random import randint
import random

header = ['SubjID', 'Decision1', 'Decision2', 'Decision3', 'HealthStatus']
data = []
d1 = 0
d2 = 0
d3 = 0
health = 0 
health_vals = [75, 80, 85, 90, 95, 100]

for i in range(1, 61):
    health = 100    
    d1 = randint(0, 5)
    d2 = randint(0, 5)
    d3 = randint(0, 5)
    health = random.choice(health_vals) 
    
    data.append([i, d1, d2, d3, health])

with open('JointEvaluation.csv', 'w', encoding='UTF8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(data)
```

```python
import pandas as pd
import numpy as np
import plotly.express as px

data = pd.read_csv("JointEvaluation.csv")
fig = px.parallel_categories(data, title="Joint Evaluation")
fig.show()
```
