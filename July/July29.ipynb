# Uses 'tictactoe_moves_1.csv', 'tictactoe_moves_2.csv', 'X_TicTacToe.csv', and 'O_TicTacToe.csv' data

```python
import pandas as pd
import numpy as np
import plotly.express as px

tictactoe_moves_1 = pd.read_csv("tictactoe_moves_1.csv", header=None)
tictactoe_moves_2 = pd.read_csv("tictactoe_moves_2.csv", header=None)

tictactoe_moves_1.fillna('', inplace=True)
tictactoe_moves_2.fillna('', inplace=True)

data = pd.concat([tictactoe_moves_1, tictactoe_moves_2])
data
```

```python
x_data = data.values.tolist()

for row in x_data:
    for x in row:
        x = str(x)
        if "O" in x:
            row.remove(x)

    while("" in row) :
        row.remove("")
        
for row2 in x_data:
    row2.append(len(row2)-1)

x_data
```

```python
o_data = data.values.tolist()

for row in o_data:
    for x in row:
        x = str(x)
        if "X" in x:
            row.remove(x)

    while("" in row) :
        row.remove("")
        
for row2 in o_data:
    row2.append(len(row2)-1)

o_data
```

```python
for row in x_data:
    for y in range(len(row)):
        row[y] = str(row[y])
        if 'X' in row[y]:
            row[y] = int(row[y][1])
        
        row[y] = float(row[y])

x_data
```

```python
for row in o_data:
    for y in range(len(row)):
        row[y] = str(row[y])
        if 'O' in row[y]:
            row[y] = int(row[y][1])
        
        row[y] = float(row[y])

o_data
```

```python
def sortLast(val):
    return val[-1]
 
def sortSecondtoLast(val):
    return val[-2]

x_data.sort(key=sortLast,reverse=True)
x_data.sort(key=sortSecondtoLast,reverse=True)

for i in x_data:
    print(i)
```

```python
def sortLast(val):
    return val[-1]
 
def sortSecondtoLast(val):
    return val[-2]

o_data.sort(key=sortLast,reverse=True)
o_data.sort(key=sortSecondtoLast,reverse=True)

for i in o_data:
    print(i)
```

```python
import csv

extra_x = 0.0
header = ['Outcome', 'Decision1', 'Decision2', 'Decision3', 'Decision4', 'Decision5']

x_final = []

for row in x_data: 
    while len(row) < 7:
        row.insert(-2, extra_x)
    
    x_final.append([row[-2], int(row[0]), int(row[1]), int(row[2]), int(row[3]), int(row[4])])


with open('X_TicTacToe.csv', 'w', encoding='UTF8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(x_final)
```
