```python
import csv
from random import randint

header = ['SubjID', 'Decision1', 'Decision2', 'Decision3', 'Decision4', 'Decision5', 'HealthStatus', 'Bias']
data = []
d1 = 0
d2 = 0
d3 = 0
d4 = 0
d5 = 0
bias = 0
health = 0 

for i in range(1, 201):
    health = 100
    
    if(i % 2 == 0):
        x = 0
    else:
        x = 1
    
    d1 = randint(0, 5)
    d2 = randint(0, 5)
    d3 = randint(0, 5)
    d4 = randint(0, 5)
    d5 = randint(0, 5)
    
    if(d1 != 5):
        health -= 5
    
    if(d2 != 5):
        health -= 5
    
    if(d3 != 5):
        health -= 5
    
    if(d4 != 5):
        health -= 5
        
    if(d5 != 5):
        health -= 5
    
    data.append([i, d1, d2, d3, d4, d5, health, x])

with open('SampleData.csv', 'w', encoding='UTF8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(data)
```

```python
import pandas as pd
import numpy as np
import plotly.express as px

data = pd.read_csv("SampleData.csv")
data
```

```python
fig = px.parallel_coordinates(
    data, 
    color="Bias", 
    labels={},
    color_continuous_scale=px.colors.diverging.Portland,
    color_continuous_midpoint=2)

fig.update_layout(coloraxis_showscale=False)
fig.show()
```
